```python
###Code to produce the tables in figure X of the paper

import matplotlib.pyplot as plt
import pandas as pd

# Input

df = pd.DataFrame({
    "Study": ["Study 1", "Study 2", "Study 3", "Study 4", "Study 5"],
    "Features": [
        "Color, Shape",
        "Color, Tail, Wings",
        "Color, Feet",
        "Stripes, Feet",
        "Stripes, Feet"
    ],
    "Participants informed of\n features up front?": ["Yes", "Yes", "No", "No", "Yes"],
    "Noise (SD)\n in DGP": [1, 1, 2, 2, 2],
    "Minimum Association of 0.5 \nfor the irrelevant feature?": ["No", "Yes", "Yes", "Yes", "Yes"],
    "Midpoint\nRatings?": ["No", "No", "No", "No", "Yes"],
    "Cognitive Process \nQuestion?": ["No", "No", "No", "Yes", "Yes"]
})

# Figure

fig, ax = plt.subplots(figsize=(13, 4))
ax.axis("off")

table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc="center",
    colLoc="center",
    loc="center",
    colWidths=[0.10, 0.16, 0.14, 0.15, 0.16, 0.14, 0.14]
)

# Styling

table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1, 1.8)

n_rows = len(df)
n_cols = len(df.columns)

for (row, col), cell in table.get_celld().items():

    #no borders
    cell.set_linewidth(0)

    if row == 0:
        # Header
        cell.set_text_props(weight="bold")
        cell.set_height(0.18)

        # Top and bottom rule
        cell.visible_edges = "TB"
        cell.set_linewidth(1.2)

    else:
        cell.set_height(0.14)

        # Bottom rule only on final row
        if row == n_rows:
            cell.visible_edges = "B"
            cell.set_linewidth(1.2)

# Left-align Study and Features
for row in range(1, n_rows + 1):
    table[row, 0].get_text().set_ha("left")
    table[row, 1].get_text().set_ha("left")

# Also left-align those headers
table[0, 0].get_text().set_ha("left")
table[0, 1].get_text().set_ha("left")

plt.tight_layout()

# Save 
#plt.savefig("study_comparison_table.pdf", bbox_inches="tight")
#plt.savefig("study_comparison_table.png", dpi=300, bbox_inches="tight")

plt.show()
```
