```python
from pathlib import Path
import pandas as pd
import shutil

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

source_dir = Path("/Users/sm6511/Downloads/data7")

destination_dir = Path(
    "/Users/sm6511/Desktop/Prediction-Accomodation-Exp/data/Study6.0Pilot/raw"
)

destination_dir.mkdir(parents=True, exist_ok=True)

# Column indicating experiment completion
completion_column = "debrief_survey.block_1/debriefComplete"


# ------------------------------------------------------------
# Loop through raw CSV files
# ------------------------------------------------------------

copied_files = []
skipped_files = []
error_files = []

for file_path in source_dir.glob("*.csv"):

    try:
        df = pd.read_csv(file_path)

        # Make sure the completion column exists
        if completion_column not in df.columns:
            print(f"Missing completion column: {file_path.name}")
            skipped_files.append(file_path.name)
            continue

        # Check whether "complete" appears anywhere in the column
        completed = (
            df[completion_column]
            .astype(str)
            .str.strip()
            .str.lower()
            .eq("complete")
            .any()
        )

        if completed:

            destination_path = destination_dir / file_path.name

            shutil.copy2(
                file_path,
                destination_path
            )

            copied_files.append(file_path.name)

            print(f"COPIED: {file_path.name}")

        else:
            skipped_files.append(file_path.name)

            print(f"SKIPPED: {file_path.name}")

    except Exception as e:
        error_files.append(file_path.name)

        print(f"ERROR reading {file_path.name}: {e}")


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n------------------------------")
print("Finished")
print("------------------------------")
print(f"Copied:  {len(copied_files)}")
print(f"Skipped: {len(skipped_files)}")
print(f"Errors:  {len(error_files)}")
```
